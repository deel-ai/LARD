```python
import pandas as pd

airports_df = pd.read_csv('airports_ourAirports_REDUCED.csv', sep=';')
runways_df = pd.read_csv('runways_FLSim_REDUCED.csv', sep=';')

common_idents = set(airports_df['ident']).intersection(set(runways_df['@ident']))

filtered_airports_df = airports_df[airports_df['ident'].isin(common_idents)]
filtered_runways_df = runways_df[runways_df['@ident'].isin(common_idents)]

filtered_airports_df.to_csv('fused_airports_ourAirports_REDUCED.csv', sep=';', index=False)
filtered_runways_df.to_csv('fused_runways_FLSim_REDUCED.csv', sep=';', index=False)
```

```python
import json
import pandas as pd

filtered_runways_df = pd.read_csv('fused_runways_FLSim_REDUCED.csv', sep=';')
icao_codes_to_keep = set(filtered_runways_df['@ident'])

with open('../full_runways_database.json', 'r') as f:
    runways_data = json.load(f)

filtered_runways_data = {icao: data for icao, data in runways_data.items() if icao in icao_codes_to_keep}

with open('../filtered_full_runways_database.json', 'w') as f:
    json.dump(filtered_runways_data, f, indent=2)
```
